```python
plt.figure(figsize=(12,10), facecolor="#c8aca9")
plt.suptitle("Customer sales Dashboard",fontsize=24,fontweight="bold",color="#658cbb")
# 1. Gender Ratio (Pie Chart)
plt.subplot(2,2,1)
gender = df["Gender"].value_counts()
plt.pie(gender,labels=gender.index,colors=["#ef4026", "#b1d1fc"],autopct="%1.1f%%",startangle=90,explode=(0.02,0),shadow=True)
plt.title("Gender Ratio",fontsize=16,fontweight="bold",color="#b7c9e2")
# 2. Gender Count (Count Plot)
plt.subplot(2,2,2)
sns.countplot(x="Gender",data=df,hue="Gender",palette=["#247afd", "#f29e8e"])
plt.gca().set_facecolor("#f4320c")
plt.title("Gender Count Ratio",fontsize=16,fontweight="bold",color="#ff6f52")
plt.xlabel("Gender")
plt.ylabel("Count")
# 3. Marital Count (Count Plot)
plt.subplot(2,2,3)
sns.countplot(x="Marital",data=df,hue="Marital",palette=["#a2bffe", "#fea993"])
plt.gca().set_facecolor("#2138ab")
plt.title("Marital Status ",fontsize=16,fontweight="bold",color="#b66a50")
plt.xlabel("Marital Status")
plt.ylabel("Count")
# 4. Marital Ratio (Donut Chart)
plt.subplot(2,2,4)
marital = df["Marital"].value_counts()
plt.pie(marital,labels=marital.index,colors=["#4b57db", "#d99b82"],autopct="%1.1f%%",startangle=90,shadow=True,wedgeprops={"width":0.45})
plt.title("Marital Ratio",fontsize=16,fontweight="bold",color="#fc824a")
plt.tight_layout()
plt.show()
```

```python
df = pd.read_csv('sales_data.csv', encoding='latin1')

# Check for null values
display(df.isnull().sum())
```

```python
# Drop unwanted columns and handle missing values
df.drop(['Status', 'unnamed1'], axis=1, inplace=True)
df.dropna(inplace=True)

# Verify data types and nulls after cleaning
df.info()
```

```python
# Rename and map values for better readability
df.rename(columns={'Marital_Status': 'Marital'}, inplace=True)

df['Marital'] = df['Marital'].map({0: 'single', 1: 'married'})
df['Gender'] = df['Gender'].map({'F': 'Female', 'M': 'Male'})

display(df.head())
```

### Data Visualization
We will now visualize the data to understand spending patterns across different categories.

```python
# 1. Gender Spending
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
ax = sns.countplot(x='Gender', data=df, hue='Gender', palette='viridis', legend=False)
plt.title('Count of Sales by Gender')
for p in ax.patches:
    ax.annotate(f'{p.get_height()}', (p.get_x() + p.get_width() / 2., p.get_height()), ha='center', va='baseline')

plt.subplot(1, 2, 2)
gender_spend = df.groupby(['Gender'], as_index=False)['Amount'].sum()
sns.barplot(x='Gender', y='Amount', data=gender_spend, hue='Gender', palette='magma', legend=False)
plt.title('Total Amount Spent by Gender')
plt.tight_layout()
plt.show()
```

```python
# 2. Age Group Spending
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
sns.countplot(data=df, x='Age Group', hue='Gender', palette='coolwarm')
plt.title('Count of Sales by Age Group & Gender')

plt.subplot(1, 2, 2)
age_spend = df.groupby(['Age Group'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)
sns.barplot(x='Age Group', y='Amount', data=age_spend, palette='rocket', hue='Age Group', legend=False)
plt.title('Total Spending by Age Group')
plt.tight_layout()
plt.show()
```

```python
# 3. State Spending (Top 10)
state_spend = df.groupby(['State'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False).head(10)
plt.figure(figsize=(15, 5))
sns.barplot(data=state_spend, x='State', y='Amount', palette='Set2', hue='State', legend=False)
plt.title('Top 10 States by Spending')
plt.xticks(rotation=45)
plt.show()
```

```python
# 4. Marital Status and Occupation
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
sns.countplot(data=df, x='Marital', hue='Gender', palette='pastel')
plt.title('Sales by Marital Status')

plt.subplot(1, 2, 2)
occ_spend = df.groupby(['Occupation'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)
sns.barplot(data=occ_spend, x='Occupation', y='Amount', hue='Occupation', palette='flare', legend=False)
plt.title('Spending by Occupation')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
# 5. Product Category Spending
prod_spend = df.groupby(['Product_Category'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False).head(10)
plt.figure(figsize=(15, 5))
sns.barplot(data=prod_spend, x='Product_Category', y='Amount', palette='viridis', hue='Product_Category', legend=False)
plt.title('Top 10 Product Categories by Spending')
plt.xticks(rotation=45)
plt.show()
```

### Customer Sales Distribution

```python
fig, axes = plt.subplots(2, 2, figsize=(10, 9), facecolor='#cafffb')
fig.suptitle('Customer Sales Distribution', fontsize=20)

# 1. Gender Distribution (Sales Share)
gender_data = df.groupby('Gender')['Amount'].sum()
axes[0, 0].pie(gender_data, labels=gender_data.index, autopct='%1.1f%%', startangle=140, colors=['#ff9999','#66b3ff'])
axes[0, 0].set_title('Gender Wise Sales Distribution (Amount)')

# 2. Gender Wise Customer Count
sns.countplot(ax=axes[0, 1], data=df, x='Gender', palette='viridis', hue='Gender', legend=False)
axes[0, 1].set_title('Customer Count by Gender')
for p in axes[0, 1].patches:
    axes[0, 1].annotate(f'{int(p.get_height())}', (p.get_x() + p.get_width() / 2., p.get_height()), ha='center', va='bottom')

# 3. Marital Status Distribution (Sales Share)
marital_data = df.groupby('Marital')['Amount'].sum()
axes[1, 0].pie(marital_data, labels=marital_data.index, autopct='%1.1f%%', startangle=140, colors=['#99ff99','#ffcc99'])
axes[1, 0].set_title('Marital Status Wise Sales Distribution (Amount)')

# 4. Marital Status Share (Customer Count)
sns.countplot(ax=axes[1, 1], data=df, x='Marital', palette='magma', hue='Marital', legend=False)
axes[1, 1].set_title('Customer Count by Marital Status')
for p in axes[1, 1].patches:
    axes[1, 1].annotate(f'{int(p.get_height())}', (p.get_x() + p.get_width() / 2., p.get_height()), ha='center', va='bottom')

plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```

### Correlation Analysis
Next, we will check the correlation between the numerical columns: Age, Orders, and Amount.

```python
# Calculate correlation matrix
corr_matrix = df[['Age', 'Orders', 'Amount']].corr()

# Create heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Correlation Heatmap: Age, Orders, and Amount')
plt.show()
```

## Summary Report: Key Insights from Sales Analysis

This report summarizes the major findings from the Exploratory Data Analysis of the customer sales dataset.

### 1. Primary Customer Profile
The data reveals that the most frequent and highest-spending customers are **Females** within the **26-35 age group**. This demographic significantly outperforms all others in both order count and total expenditure.

### 2. Geographic Hotspots
Sales are concentrated in specific regions, with **Uttar Pradesh** emerging as the leading state by a wide margin, followed by **Maharashtra** and **Karnataka**. Marketing efforts should prioritize these high-growth zones.

### 3. Occupational Influence
Customers working in the **IT Sector, Healthcare, and Aviation** show the highest purchasing power. This suggests that professionals in high-growth industries are the primary drivers of sales volume.

### 4. Top Product Categories
**Food** is the dominant product category by total sales amount, nearly doubling the spending seen in the next categories, **Clothing & Apparel** and **Electronics & Gadgets**.

### 5. Social Dynamics
**Single individuals** account for a larger share of the sales (58.5%) compared to married customers (41.5%), indicating that independent consumers are a vital segment for this business.

**Strategic Recommendation:** Target digital marketing campaigns toward single women aged 26-35 in Tier-1 cities of Uttar Pradesh and Maharashtra, focusing on Food and Apparel promotions.
